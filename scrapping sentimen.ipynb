```python
from google_play_scraper import app, reviews_all, Sort
import pandas as pd
import numpy as np

pd.options.mode.chained_assignment = None
seed = 0
np.random.seed(seed)

print('Library berhasil di-import!')
```

```python
scrapreview = reviews_all(
    'com.apps.MyXL',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=1000
)

print(f'Berhasil scraping {len(scrapreview)} review')
```

```python
# Konversi hasil scraping ke DataFrame
app_reviews_df = pd.DataFrame(scrapreview)

jumlah_ulasan, jumlah_kolom = app_reviews_df.shape
print(f'Jumlah ulasan: {jumlah_ulasan}')
print(f'Jumlah kolom: {jumlah_kolom}')
print()

app_reviews_df.head()
```

```python
# Cek duplikat sebelum dihapus
jumlah_duplikat = app_reviews_df.duplicated().sum()
print(f'Jumlah data duplikat: {jumlah_duplikat}')

# Hapus duplikat
app_reviews_df = app_reviews_df.drop_duplicates()

jumlah_setelah, kolom_setelah = app_reviews_df.shape
print(f'Jumlah ulasan setelah hapus duplikat: {jumlah_setelah}')
```

```python
# Simpan seluruh kolom ke CSV
app_reviews_df.to_csv('komen_XL.csv', index=False)

print(f'Data berhasil disimpan ke komen_XL.csv ({len(app_reviews_df)} baris)')
print(f'Kolom yang tersimpan: {list(app_reviews_df.columns)}')
```
